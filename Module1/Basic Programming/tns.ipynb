```python
age = int(input("Please enter your age: "))

if age >= 18:
    print("You are eligible to vote!")
elif age >= 16:
    print("You are almost eligible to vote, come back in a few years!")
else:
    print("You are not old enough to vote yet.")
```

```python
# 1. len(): Returns the length (the number of items) of an object.
my_list = [10, 20, 30, 40, 50]
my_string = "Hello, Python!"
print(f"Length of list: {len(my_list)}")
print(f"Length of string: {len(my_string)}")

# 2. type(): Returns the type of an object.
my_variable = 123
print(f"Type of my_variable: {type(my_variable)}")
print(f"Type of my_list: {type(my_list)}")

# 3. print(): Prints objects to the text stream file, separated by sep and followed by end.
print("This is a message.")
print("Multiple", "arguments", "here.", sep='-', end='!\n')

# 4. max(): Returns the largest item in an iterable or the largest of two or more arguments.
numbers = [5, 1, 9, 3, 7]
print(f"Maximum number in list: {max(numbers)}")
print(f"Maximum of two numbers (10, 20): {max(10, 20)}")

# 5. min(): Returns the smallest item in an iterable or the smallest of two or more arguments.
print(f"Minimum number in list: {min(numbers)}")
print(f"Minimum of two numbers (10, 20): {min(10, 20)}")

# 6. sum(): Sums start and the items of an iterable from left to right and returns the total.
print(f"Sum of numbers in list: {sum(numbers)}")

# 7. abs(): Returns the absolute value of a number.
negative_number = -15
print(f"Absolute value of {negative_number}: {abs(negative_number)}")

# 8. round(): Rounds a number to a given precision in decimal digits (default 0 decimal places).
float_number = 3.14159
print(f"Rounded {float_number} to 2 decimal places: {round(float_number, 2)}")
print(f"Rounded {float_number} to nearest integer: {round(float_number)}")
```

```python
import sys
import os

# 1. Create the package directory and files
# Ensure the directory exists. -p flag creates parent directories if they don't exist.
!mkdir -p my_custom_package

# Create an empty __init__.py file inside the package to make it a Python package
# Using a temporary file for %%writefile output to ensure proper execution within one cell
with open('my_custom_package/__init__.py', 'w') as f:
    f.write('# This file makes \'my_custom_package\' a Python package\n')

# Create the module file 'utils.py' inside 'my_custom_package'
with open('my_custom_package/utils.py', 'w') as f:
    f.write('def greet(name):\n')
    f.write('    """Greets the given name."""\n')
    f.write('    return f"Hello, {name}!"\n\n')
    f.write('def add(a, b):\n')
    f.write('    """Adds two numbers."""\n')
    f.write('    return a + b\n\n')
    f.write('def multiply(a, b):\n')
    f.write('    """Multiplies two numbers."""\n')
    f.write('    return a * b\n')


# 2. Add the current directory to the Python path
# This allows Python to find 'my_custom_package'
if '.' not in sys.path:
    sys.path.insert(0, '.') # Insert at the beginning to prioritize local packages

# 3. Import the module from the package
# Now that the path is set and files are created, this import should work
from my_custom_package import utils

# 4. Access and use the functions
print(f"\n--- Using functions from my_custom_package.utils ---")
print(utils.greet("Charlie"))
print(f"Sum of 10 and 5: {utils.add(10, 5)}")
print(f"Product of 7 and 8: {utils.multiply(7, 8)}")

# You can also import specific functions directly
from my_custom_package.utils import greet, add

print(f"\n--- Using directly imported functions ---")
print(greet("Diana"))
print(f"Sum of 100 and 200: {add(100, 200)}")

# Optional: Clean up the created directory after use
# import shutil
# shutil.rmtree('my_custom_package')
```
